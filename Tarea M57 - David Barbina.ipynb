```python
df = pd.read_csv("Population_Surface_Area_and_Density.csv", encoding="latin1", skiprows=1, header=0)
df
```

```python
df = df.rename(columns={"Unnamed: 1": "Area"})
df
```

```python
df["Value"] = df["Value"].astype(str).str.replace(",", "").astype(float)
```

```python
df_poblacion = df[df["Series"] == "Population mid-year estimates (millions)"]
df_poblacion
```

```python
df_poblacion_2010 = df[(df["Series"] == "Population mid-year estimates (millions)") &
    (df["Year"] == 2010)]
df_poblacion_2010 = df_poblacion_2010.iloc[1:]
df_poblacion_2010
```

```python
df_poblacion_2025 = df[(df["Series"] == "Population mid-year estimates (millions)") &
    (df["Year"] == 2025)]

df_poblacion_2025 = df_poblacion_2025.iloc[1:]
df_poblacion_2025
```

```python
# Se despliega un gráfico de marcadores y precisamente se elijen marcadores gracias a los consejos de Edward Tufte sobre la eliminación de
# elementos innecesarios en gráficos para mejorar la claridad y la efectividad de la comunicación visual es y para seguir el consejo de Nigel 
# Holmes para evitar la saturación visual se opta por restringir a 8 registros.


df_top8 = df_poblacion_2010.head(8)

fig1 = go.Figure()
fig1.add_trace(go.Scatter(x=df_top8["Area"], y=df_top8["Value"], mode="markers", name="Población", line=dict(color="red")))
fig1.update_layout(title="Población por región/área (2025)", xaxis_title="Región/Área", yaxis_title="Población (millones)")
fig1.show()
```

```python
# Gráfico de barras para la población por región/área (2025)

df_top8_2025 = df_poblacion_2025.head(8)

fig3 = go.Figure()
fig3.add_trace(go.Bar(x=df_top8_2025["Area"], y=df_top8_2025["Value"], name="Población", marker=dict(color="green")))
fig3.update_layout(title="Población por región/área (2025)", xaxis_title="Región/Área", yaxis_title="Población (millones)")
fig3.show()
```

```python
import dash
from dash import dcc
from dash import html

# Consolidación de gráficos

app = dash.Dash(__name__)
app.layout = html.Div([
    html.H1('Cuadro de Control de Mando (Dashboard)'),

    html.Div([
        dcc.Graph(figure=fig1),
    ], style={'display': 'flex'}),

    html.Div([
        dcc.Graph(figure=fig3),
    ], style={'display': 'flex'}),
])
if __name__ == '__main__':
    app.run(debug=True)
```
